# Dataset 1 - Appliances Energy Prediction

**Situação:** uma empresa de eficiência energética está analisando o comportamento de uma residência de baixo consumo. A equipe deseja identificar períodos de consumo elevado dos eletrodomésticos e observar quais condições de temperatura e umidade estavam presentes nesses momentos.

## 1. Carregamento e inspeção inicial da amostra

```python
from pathlib import Path
import pandas as pd

nomes_dataset1 = [
    'energydata_SAMPLE_01.csv'
]
pastas_busca_1 = [
    Path('/content'),
    Path.cwd(),
    Path.home() / 'Downloads',
    Path.home() / 'Documents' / 'arquivos-cp-sers'
]
arquivo_dataset1 = next(
    (
        pasta / nome
        for pasta in pastas_busca_1
        for nome in nomes_dataset1
        if (pasta / nome).exists()
    ),
    None
)

if arquivo_dataset1 is None:
    raise FileNotFoundError(
        'Envie o arquivo DATASET-1.csv para o Colab antes de executar.'
    )

dados1 = pd.read_csv(arquivo_dataset1)
inicio_dataset1 = ' '.join(
    map(str, dados1.head(2).to_numpy().ravel())
).lower()
if 'continuous' in inicio_dataset1 or 'meta' in inicio_dataset1:
    dados1 = pd.read_csv(arquivo_dataset1, skiprows=[1, 2])

colunas_dataset1 = [
    'Appliances', 'lights', 'T1', 'RH_1', 'T2', 'RH_2', 'T3', 'RH_3'
]
dados1 = dados1[colunas_dataset1].copy()

print(f'Dimensões da amostra: {dados1.shape[0]} linhas e {dados1.shape[1]} colunas')
dados1.head()
```

```python
dados1.info()
print()
print('Valores ausentes por coluna:')
display(dados1.isnull().sum().to_frame('Quantidade'))
dados1.describe()
```

## 2. Renomeação dos atributos

```python
dados1.rename(columns={
    'Appliances': 'Consumo_Eletrodomesticos',
    'lights': 'Iluminacao',
    'T1': 'Temperatura_1',
    'RH_1': 'Umidade_1',
    'T2': 'Temperatura_2',
    'RH_2': 'Umidade_2',
    'T3': 'Temperatura_3',
    'RH_3': 'Umidade_3'
}, inplace=True)

print(dados1.columns.tolist())
dados1.head()
```

## 3. Maior consumo de eletrodomésticos

```python
consumo_maximo_1 = dados1['Consumo_Eletrodomesticos'].max()
print(f'Maior consumo registrado: {consumo_maximo_1:.2f} Wh')
```

## 4 e 5. Limiar de 70%, quantidade e percentual

```python
limiar_70_1 = 0.70 * consumo_maximo_1
df_alto_consumo_1 = dados1[
    dados1['Consumo_Eletrodomesticos'] > limiar_70_1
].copy()

quantidade_alto_1 = len(df_alto_consumo_1)
percentual_alto_1 = quantidade_alto_1 / len(dados1) * 100

print(f'Limiar de 70% do máximo: {limiar_70_1:.2f} Wh')
print(f'Registros acima do limiar: {quantidade_alto_1}')
print(f'Percentual da amostra: {percentual_alto_1:.2f}%')
df_alto_consumo_1.sort_values(
    'Consumo_Eletrodomesticos', ascending=False
).head(10)
```

## 6. Alto consumo e temperatura acima da média

```python
temperatura_media_1 = dados1['Temperatura_1'].mean()
df_alto_consumo_temperatura_1 = dados1[
    (dados1['Consumo_Eletrodomesticos'] > limiar_70_1) &
    (dados1['Temperatura_1'] > temperatura_media_1)
].copy()

quantidade_duplo_1 = len(df_alto_consumo_temperatura_1)
percentual_dentro_altos_1 = quantidade_duplo_1 / quantidade_alto_1 * 100

print(f'Temperatura média de T1: {temperatura_media_1:.2f} °C')
print(f'Registros com as duas condições: {quantidade_duplo_1}')
print(f'Percentual entre os casos de alto consumo: {percentual_dentro_altos_1:.2f}%')
df_alto_consumo_temperatura_1.head(10)
```

## 7. Comparação e interpretação

```python
from IPython.display import Markdown, display

reducao_1 = quantidade_alto_1 - quantidade_duplo_1
reducao_percentual_1 = reducao_1 / quantidade_alto_1 * 100

display(Markdown(f'''
**Interpretação:** o filtro somente por consumo selecionou **{quantidade_alto_1} registros**. Ao incluir a temperatura acima da média, o total caiu para **{quantidade_duplo_1} registros**, redução de **{reducao_1} registros ({reducao_percentual_1:.2f}%)**.

Isso acontece porque o segundo DataFrame exige duas condições simultâneas e, portanto, contém apenas a parte dos episódios de alto consumo que também ocorreu com temperatura acima da média. A relação observada descreve a amostra, mas não prova que a temperatura causou o consumo elevado.
'''))
```

# Dataset 2 - Steel Industry Energy Consumption

**Situação:** a equipe de gestão de energia de uma indústria siderúrgica deseja localizar situações de consumo elevado e verificar se esses períodos coincidem com condições de carga elevada ou valores menos favoráveis de fator de potência.

A análise utiliza a amostra aleatória de 20% preparada no Orange Data Mining.

## 1. Carregamento da amostra e renomeação das colunas

O Orange acrescentou duas linhas de metadados ao CSV. Por isso, elas são ignoradas com `skiprows=[1, 2]`. Se o arquivo não estiver na pasta do notebook ou em Downloads, o código permite enviá-lo pelo Google Colab.

```python
import pandas as pd
from pathlib import Path

nomes_dataset2 = [
    'Steel_industry_data_amostra_20_percent.csv'
]
pastas_busca_2 = [
    Path('/content'),
    Path.cwd(),
    Path.home() / 'Downloads',
    Path.home() / 'Documents' / 'arquivos-cp-sers'
]
arquivo = next(
    (
        pasta / nome
        for pasta in pastas_busca_2
        for nome in nomes_dataset2
        if (pasta / nome).exists()
    ),
    None
)

if arquivo is None:
    raise FileNotFoundError(
        'Envie o arquivo DATASET-2.csv para o Colab antes de executar.'
    )

dados = pd.read_csv(arquivo)
inicio_dataset2 = ' '.join(
    map(str, dados.head(2).to_numpy().ravel())
).lower()
if 'continuous' in inicio_dataset2 or 'meta' in inicio_dataset2:
    dados = pd.read_csv(arquivo, skiprows=[1, 2])

dados.rename(columns={
    'date': 'Data_Hora',
    'Usage_kWh': 'Consumo_kWh',
    'Leading_Current_Power_Factor': 'Fator_Potencia_Adiantado',
    'Lagging_Current_Power_Factor': 'Fator_Potencia_Atrasado',
    'Leading_Current_Reactive_Power_kVarh': 'Pot_Reativa_Adiantada_kVArh',
    'Lagging_Current_Reactive.Power_kVarh': 'Pot_Reativa_Atrasada_kVArh',
    'WeekStatus': 'Status_Semana',
    'Day_of_week': 'Dia_Semana',
    'Load_Type': 'Tipo_Carga'
}, inplace=True)

dados['Data_Hora'] = pd.to_datetime(
    dados['Data_Hora'], format='%d/%m/%Y %H:%M', errors='coerce'
)
print('Colunas após a renomeação:')
print(dados.columns.tolist())
```

## 2. Inspeção inicial: `head()`, `shape`, `info()` e `describe()`

```python
dados.head()
```

```python
print(f'Dimensões da amostra: {dados.shape[0]} linhas e {dados.shape[1]} colunas')
dados.shape
```

```python
dados.info()
```

```python
dados.describe(include='all')
```

### Verificação complementar da qualidade e das categorias

```python
print('Valores ausentes por coluna:')
display(dados.isnull().sum().to_frame('Quantidade'))

print('Categorias de Tipo_Carga:')
display(dados['Tipo_Carga'].value_counts().to_frame('Quantidade'))

print('Categorias de Status_Semana:')
display(dados['Status_Semana'].value_counts().to_frame('Quantidade'))
```

## 3. Consumo máximo e limiar de 75%

```python
consumo_maximo = dados['Consumo_kWh'].max()
limiar_75 = 0.75 * consumo_maximo

print(f'Maior consumo registrado: {consumo_maximo:.2f} kWh')
print(f'Limiar de 75% do consumo máximo: {limiar_75:.2f} kWh')
```

## 4. DataFrame com registros acima do limiar

```python
df_alto_consumo = dados[dados['Consumo_kWh'] > limiar_75].copy()
quantidade_alto_consumo = len(df_alto_consumo)
percentual_alto_consumo = quantidade_alto_consumo / len(dados) * 100

print(f'Quantidade de registros de alto consumo: {quantidade_alto_consumo}')
print(f'Percentual da amostra: {percentual_alto_consumo:.2f}%')
df_alto_consumo.sort_values('Consumo_kWh', ascending=False).head(10)
```

## 5. Registros de alto consumo classificados como `Maximum_Load`

```python
contagem_cargas_alto_consumo = df_alto_consumo['Tipo_Carga'].value_counts()
quantidade_maximum_load = (df_alto_consumo['Tipo_Carga'] == 'Maximum_Load').sum()
percentual_maximum_no_alto = quantidade_maximum_load / len(df_alto_consumo) * 100

display(contagem_cargas_alto_consumo.to_frame('Quantidade'))
print(f'Registros de alto consumo em Maximum_Load: {quantidade_maximum_load}')
print(f'Percentual dentro do DataFrame de alto consumo: {percentual_maximum_no_alto:.2f}%')
```

**Interpretação:** a classificação `Maximum_Load` aparece em parte importante dos registros de consumo elevado, mas não em todos. Isso acontece porque o tipo de carga é uma categoria operacional, enquanto o filtro foi definido diretamente pelos valores de consumo observados na amostra.

## 6. Escolha e observação do fator de potência

Foi escolhido o fator de potência atrasado (`Fator_Potencia_Atrasado`). Para esta análise exploratória, valores abaixo de **90%** serão considerados mais baixos. O limite é coerente porque o fator de potência se aproxima de uma condição mais favorável quando está perto de 100%, e o próprio recorte de alto consumo apresenta valores aproximadamente entre 84% e 95%.

```python
coluna_fator = 'Fator_Potencia_Atrasado'
limite_fator = 90.0

print('Distribuição do fator de potência atrasado na amostra completa:')
display(dados[coluna_fator].describe(percentiles=[0.10, 0.25, 0.50, 0.75, 0.90]).to_frame())

print('Distribuição do fator de potência atrasado nos registros de alto consumo:')
display(df_alto_consumo[coluna_fator].describe().to_frame())

print(f'Limite escolhido para representar valores mais baixos: {limite_fator:.2f}%')
```

## 7. Consumo elevado e fator de potência abaixo do limite

```python
df_alto_consumo_fator_baixo = dados[
    (dados['Consumo_kWh'] > limiar_75) &
    (dados[coluna_fator] < limite_fator)
].copy()

quantidade_critica = len(df_alto_consumo_fator_baixo)
percentual_amostra_critica = quantidade_critica / len(dados) * 100
percentual_dos_altos = quantidade_critica / len(df_alto_consumo) * 100

print(f'Registros com alto consumo e fator de potência abaixo de {limite_fator:.0f}%: {quantidade_critica}')
print(f'Percentual da amostra completa: {percentual_amostra_critica:.2f}%')
print(f'Percentual entre os registros de alto consumo: {percentual_dos_altos:.2f}%')
df_alto_consumo_fator_baixo.sort_values(
    ['Fator_Potencia_Atrasado', 'Consumo_kWh'],
    ascending=[True, False]
).head(10)
```

## 8. Interpretação final

O segundo DataFrame merece maior atenção porque reúne duas condições ao mesmo tempo: consumo superior a 75% do máximo da amostra e fator de potência atrasado inferior a 90%. Nesses períodos, a instalação apresenta demanda energética elevada junto de uma condição de fator de potência menos favorável, o que pode estar associado a maior circulação de potência reativa, perdas e uso menos eficiente da infraestrutura elétrica.

Na amostra analisada, **129 registros (1,84%)** ficaram acima do limiar de consumo. Desses, **62** foram classificados como `Maximum_Load`. O filtro combinado selecionou **42 registros**, equivalentes a **0,60% da amostra completa** e **32,56% dos registros de alto consumo**.

Esse recorte ajuda a equipe a priorizar períodos para investigação, mas não comprova sozinho a existência de falha. Uma conclusão operacional exigiria também avaliar equipamentos ligados, processo produtivo, potência reativa, histórico e critérios técnicos da instalação.

# Dataset 3 - Power Consumption of Tetouan City

**Situação:** uma equipe responsável pelo planejamento da distribuição de energia em Tétouan deseja identificar qual das três zonas apresenta o maior pico de consumo e observar as condições ambientais registradas nos momentos de maior demanda.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Markdown, display

nome_dataset3 = 'Tetuan_City_power_consumption_15_percent.csv'
candidatos_dataset3 = [
    Path('/content') / nome_dataset3,
    Path.cwd() / nome_dataset3,
    Path.home() / 'Downloads' / nome_dataset3,
    Path.home() / 'Documents' / 'arquivos-cp-sers' / nome_dataset3
]
arquivo_dataset3 = next(
    (caminho for caminho in candidatos_dataset3 if caminho.exists()),
    None
)

if arquivo_dataset3 is None:
    raise FileNotFoundError(
        'Envie o arquivo DATASET-3.csv para o Colab antes de executar.'
    )

# O Orange pode acrescentar duas linhas de metadados ao CSV.
dados3 = pd.read_csv(arquivo_dataset3)
inicio_dataset3 = ' '.join(
    map(str, dados3.head(2).to_numpy().ravel())
).lower()
if 'continuous' in inicio_dataset3 or 'meta' in inicio_dataset3:
    dados3 = pd.read_csv(arquivo_dataset3, skiprows=[1, 2])

# Algumas versões do arquivo possuem espaços duplicados em Zone 2 e Zone 3.
dados3.columns = [
    ' '.join(str(coluna).strip().split())
    for coluna in dados3.columns
]

colunas_obrigatorias_3 = [
    'Temperature',
    'Humidity',
    'Wind Speed',
    'Zone 1 Power Consumption',
    'Zone 2 Power Consumption',
    'Zone 3 Power Consumption'
]
colunas_faltantes_3 = [
    coluna for coluna in colunas_obrigatorias_3
    if coluna not in dados3.columns
]

if colunas_faltantes_3:
    raise ValueError(
        'O DATASET-3.csv está incompleto. Exporte novamente pelo Orange '
        'incluindo estas colunas: ' + ', '.join(colunas_faltantes_3)
    )

dados3.rename(columns={
    'DateTime': 'Data_Hora',
    'Temperature': 'Temperatura',
    'Humidity': 'Umidade',
    'Wind Speed': 'Velocidade_Vento',
    'Zone 1 Power Consumption': 'Consumo_Zona_1',
    'Zone 2 Power Consumption': 'Consumo_Zona_2',
    'Zone 3 Power Consumption': 'Consumo_Zona_3'
}, inplace=True)

colunas_numericas_3 = [
    'Temperatura', 'Umidade', 'Velocidade_Vento',
    'Consumo_Zona_1', 'Consumo_Zona_2', 'Consumo_Zona_3'
]
for coluna in colunas_numericas_3:
    dados3[coluna] = pd.to_numeric(dados3[coluna], errors='coerce')

if 'Data_Hora' in dados3.columns:
    dados3['Data_Hora'] = pd.to_datetime(
        dados3['Data_Hora'], errors='coerce', format='mixed'
    )

print(f'Dimensões da amostra: {dados3.shape[0]} linhas e {dados3.shape[1]} colunas')
print('Colunas após a renomeação:')
print(dados3.columns.tolist())
dados3.head()
```

### Inspeção complementar da amostra

```python
dados3.info()
print()
print('Valores ausentes por coluna:')
display(dados3.isnull().sum().to_frame('Quantidade'))
dados3.describe(include='all')
```

## 2 e 3. Consumo máximo de cada zona e identificação do maior pico

```python
colunas_zonas_3 = [
    'Consumo_Zona_1',
    'Consumo_Zona_2',
    'Consumo_Zona_3'
]

maximos_zonas_3 = dados3[colunas_zonas_3].max()
zona_maior_pico_3 = maximos_zonas_3.idxmax()
maior_pico_3 = maximos_zonas_3.max()

display(maximos_zonas_3.to_frame('Consumo máximo'))
print(f'Zona com o maior pico: {zona_maior_pico_3}')
print(f'Maior pico registrado: {maior_pico_3:.2f}')
```

## 4 e 5. Limiar de 70%, quantidade e percentual de registros

```python
limiar_70_3 = 0.70 * maior_pico_3
df_alto_consumo_3 = dados3[
    dados3[zona_maior_pico_3] > limiar_70_3
].copy()

quantidade_alto_3 = len(df_alto_consumo_3)
percentual_alto_3 = quantidade_alto_3 / len(dados3) * 100

print(f'Limiar de 70% do maior pico: {limiar_70_3:.2f}')
print(f'Registros acima do limiar: {quantidade_alto_3}')
print(f'Percentual da amostra: {percentual_alto_3:.2f}%')
df_alto_consumo_3.sort_values(
    zona_maior_pico_3, ascending=False
).head(10)
```

## 6. Alto consumo e temperatura acima da média

```python
temperatura_media_3 = dados3['Temperatura'].mean()
df_alto_consumo_temperatura_3 = dados3[
    (dados3[zona_maior_pico_3] > limiar_70_3) &
    (dados3['Temperatura'] > temperatura_media_3)
].copy()

quantidade_duplo_3 = len(df_alto_consumo_temperatura_3)
percentual_duplo_3 = quantidade_duplo_3 / len(dados3) * 100

print(f'Temperatura média da amostra: {temperatura_media_3:.2f} °C')
print(f'Registros com as duas condições: {quantidade_duplo_3}')
print(f'Percentual da amostra: {percentual_duplo_3:.2f}%')
df_alto_consumo_temperatura_3.head(10)
```

## 7. Comparação dos DataFrames e interpretação

```python
reducao_3 = quantidade_alto_3 - quantidade_duplo_3
reducao_percentual_3 = (
    reducao_3 / quantidade_alto_3 * 100
    if quantidade_alto_3 > 0 else 0
)

display(Markdown(f'''
**Interpretação:** o primeiro DataFrame, que considera somente o consumo acima de 70% do maior pico da **{zona_maior_pico_3}**, possui **{quantidade_alto_3} registros**. Depois da inclusão da temperatura acima da média, o segundo DataFrame ficou com **{quantidade_duplo_3} registros**.

A segunda condição reduziu o conjunto em **{reducao_3} registros ({reducao_percentual_3:.2f}%)**, pois manteve somente os momentos de alta demanda que também ocorreram com temperatura acima da média. Essa comparação descreve a amostra analisada, mas não demonstra que a temperatura causou o aumento do consumo.
'''))
```

# Dataset 4 - Solar Power Generation Data

```python
from pathlib import Path
import pandas as pd

nomes_dataset4 = [
    'DATASET-4.csv',
    'Plant_1_Generation_Data_20_percent.csv'
]
pastas_busca_4 = [
    Path('/content'),
    Path.cwd(),
    Path.home() / 'Downloads',
    Path.home() / 'Documents' / 'arquivos-cp-sers'
]
arquivo_dataset4 = next(
    (
        pasta / nome
        for pasta in pastas_busca_4
        for nome in nomes_dataset4
        if (pasta / nome).exists()
    ),
    None
)

if arquivo_dataset4 is None:
    raise FileNotFoundError(
        'Envie o arquivo DATASET-4.csv para o Colab antes de executar.'
    )

dados4_brutos = pd.read_csv(arquivo_dataset4)
inicio_dataset4 = ' '.join(
    map(str, dados4_brutos.head(2).to_numpy().ravel())
).lower()
if 'continuous' in inicio_dataset4 or 'meta' in inicio_dataset4:
    dados4_brutos = pd.read_csv(arquivo_dataset4, skiprows=[1, 2])

colunas_dataset4 = [
    'DATE_TIME', 'SOURCE_KEY', 'DC_POWER',
    'AC_POWER', 'DAILY_YIELD', 'TOTAL_YIELD'
]
dados4 = dados4_brutos[colunas_dataset4].copy()

dados4.rename(columns={
    'DATE_TIME': 'Data_Hora',
    'DC_POWER': 'Potencia_CC',
    'AC_POWER': 'Potencia_CA',
    'DAILY_YIELD': 'Geracao_Diaria',
    'TOTAL_YIELD': 'Geracao_Total'
}, inplace=True)

for coluna in ['Potencia_CC', 'Potencia_CA', 'Geracao_Diaria', 'Geracao_Total']:
    dados4[coluna] = pd.to_numeric(dados4[coluna], errors='coerce')
dados4.dropna(inplace=True)
dados4.reset_index(drop=True, inplace=True)

print(f'Dimensões da amostra: {dados4.shape[0]} linhas e {dados4.shape[1]} colunas')
dados4.head()
```

```python
potencia_ca_maxima = dados4['Potencia_CA'].max()
limiar_70_ca = 0.70 * potencia_ca_maxima

print(f'Maior potência CA registrada: {potencia_ca_maxima:.2f}')
print(f'Limiar de 70% da potência CA máxima: {limiar_70_ca:.2f}')
```

## 5. Registros de alta geração

```python
df_alta_geracao_4 = dados4[
    dados4['Potencia_CA'] > limiar_70_ca
].copy()

quantidade_alta_4 = len(df_alta_geracao_4)
percentual_alta_4 = quantidade_alta_4 / len(dados4) * 100

print(f'Registros acima do limiar: {quantidade_alta_4}')
print(f'Percentual da amostra: {percentual_alta_4:.2f}%')
df_alta_geracao_4.sort_values(
    'Potencia_CA', ascending=False
).head(10)
```

```python
display(Markdown(f'''
**Interpretação:** o inversor **{inversor_frequente_4}** apareceu com maior frequência no recorte de alta geração, com **{quantidade_inversor_4} registros**. Isso permite observar sua presença relativa nos momentos acima de 70% da potência CA máxima da amostra.

O resultado, isoladamente, não demonstra que esse inversor possui desempenho superior nem que os demais apresentam falha. Uma conclusão desse tipo exigiria comparar disponibilidade, irradiância, horários de operação, quantidade total de registros e condições técnicas de cada equipamento.
'''))
```

# Dataset 5 - Wind & Solar Energy Production

O arquivo está no formato longo: cada linha identifica o tipo de geração e o valor produzido. Por isso, a análise separa os registros solares e eólicos antes de calcular os máximos e os limiares.

## 1. Carregamento e inspeção da amostra fornecida

```python
from pathlib import Path
import pandas as pd

nomes_dataset5 = [
    'Energy_Production_Dataset_20_percent.csv'
]
pastas_busca_5 = [
    Path('/content'),
    Path.cwd(),
    Path.home() / 'Downloads',
    Path.home() / 'Documents' / 'arquivos-cp-sers'
]
arquivo_dataset5 = next(
    (
        pasta / nome
        for pasta in pastas_busca_5
        for nome in nomes_dataset5
        if (pasta / nome).exists()
    ),
    None
)

if arquivo_dataset5 is None:
    raise FileNotFoundError(
        'Envie o arquivo DATASET-5.csv para o Colab antes de executar.'
    )

dados5 = pd.read_csv(arquivo_dataset5)
inicio_dataset5 = ' '.join(
    map(str, dados5.head(2).to_numpy().ravel())
).lower()
if 'continuous' in inicio_dataset5 or 'meta' in inicio_dataset5:
    dados5 = pd.read_csv(arquivo_dataset5, skiprows=[1, 2])

dados5.rename(columns={'Source': 'Tipo_Geracao'}, inplace=True)
dados5['Production'] = pd.to_numeric(
    dados5['Production'], errors='coerce'
)
for coluna in ['Day_of_Year', 'Start_Hour', 'End_Hour']:
    dados5[coluna] = pd.to_numeric(dados5[coluna], errors='coerce')

dados5.dropna(subset=['Tipo_Geracao', 'Production'], inplace=True)
dados5.reset_index(drop=True, inplace=True)

print(f'Dimensões da amostra: {dados5.shape[0]} linhas e {dados5.shape[1]} colunas')
dados5.head()
```

```python
dados5.info()
print()
print('Valores ausentes por coluna:')
display(dados5.isnull().sum().to_frame('Quantidade'))
print()
print('Registros por tipo de geração:')
display(dados5['Tipo_Geracao'].value_counts().to_frame('Quantidade'))
dados5.describe(include='all')
```

## 2. Separação das produções solar e eólica

```python
dados5_solar = dados5[
    dados5['Tipo_Geracao'].str.strip().str.lower() == 'solar'
].copy()
dados5_eolica = dados5[
    dados5['Tipo_Geracao'].str.strip().str.lower() == 'wind'
].copy()

dados5_solar.rename(
    columns={'Production': 'Geracao_Solar'}, inplace=True
)
dados5_eolica.rename(
    columns={'Production': 'Geracao_Eolica'}, inplace=True
)

print(f'Registros solares: {len(dados5_solar)}')
print(f'Registros eólicos: {len(dados5_eolica)}')
display(dados5_solar.head())
display(dados5_eolica.head())
```

## 3. Máximos e limiares próprios de 70%

```python
maximo_solar_5 = dados5_solar['Geracao_Solar'].max()
maximo_eolico_5 = dados5_eolica['Geracao_Eolica'].max()
limiar_solar_5 = 0.70 * maximo_solar_5
limiar_eolico_5 = 0.70 * maximo_eolico_5

print(f'Máximo solar: {maximo_solar_5:.2f}')
print(f'Limiar solar de 70%: {limiar_solar_5:.2f}')
print(f'Máximo eólico: {maximo_eolico_5:.2f}')
print(f'Limiar eólico de 70%: {limiar_eolico_5:.2f}')
```

## 4 e 5. DataFrames de alta geração, quantidades e percentuais

```python
df_alta_solar_5 = dados5_solar[
    dados5_solar['Geracao_Solar'] > limiar_solar_5
].copy()
df_alta_eolica_5 = dados5_eolica[
    dados5_eolica['Geracao_Eolica'] > limiar_eolico_5
].copy()

quantidade_solar_5 = len(df_alta_solar_5)
quantidade_eolica_5 = len(df_alta_eolica_5)
percentual_solar_5 = quantidade_solar_5 / len(dados5_solar) * 100
percentual_eolica_5 = quantidade_eolica_5 / len(dados5_eolica) * 100

print(
    f'Alta geração solar: {quantidade_solar_5} registros '
    f'({percentual_solar_5:.2f}% dos registros solares)'
)
print(
    f'Alta geração eólica: {quantidade_eolica_5} registros '
    f'({percentual_eolica_5:.2f}% dos registros eólicos)'
)

display(
    df_alta_solar_5.sort_values(
        'Geracao_Solar', ascending=False
    ).head(10)
)
display(
    df_alta_eolica_5.sort_values(
        'Geracao_Eolica', ascending=False
    ).head(10)
)
```

## 6 e 7. Comparação e interpretação

```python
from IPython.display import Markdown, display

if percentual_solar_5 > percentual_eolica_5:
    geracao_frequente_5 = 'solar'
elif percentual_eolica_5 > percentual_solar_5:
    geracao_frequente_5 = 'eólica'
else:
    geracao_frequente_5 = 'solar e eólica, com a mesma frequência'

display(Markdown(f'''
**Interpretação:** a geração **{geracao_frequente_5}** apareceu com maior frequência acima de 70% do seu próprio máximo. O recorte solar correspondeu a **{percentual_solar_5:.2f}% dos registros solares**, enquanto o recorte eólico correspondeu a **{percentual_eolica_5:.2f}% dos registros eólicos**.

Não seria adequado aplicar o mesmo limite numérico aos dois tipos de geração porque suas escalas são diferentes. O máximo solar foi **{maximo_solar_5:.2f}**, enquanto o máximo eólico foi **{maximo_eolico_5:.2f}**. O uso de 70% do máximo de cada tipo de geração permite uma comparação relativa mais coerente.
'''))
```

# Dataset 6 - Individual Household Electric Power Consumption

**Situação:** uma residência possui monitoramento elétrico detalhado e deseja identificar episódios de demanda elevada que também apresentem corrente acima do comportamento médio.

## 1. Carregamento, tratamento e organização dos atributos

```python
from pathlib import Path
import pandas as pd

nomes_dataset6 = [
    'household_power_consumption_10_percent.csv'
]
pastas_busca_6 = [
    Path('/content'),
    Path.cwd(),
    Path.home() / 'Downloads',
    Path.home() / 'Documents' / 'arquivos-cp-sers'
]
arquivo_dataset6 = next(
    (
        pasta / nome
        for pasta in pastas_busca_6
        for nome in nomes_dataset6
        if (pasta / nome).exists()
    ),
    None
)

if arquivo_dataset6 is None:
    raise FileNotFoundError(
        'Envie o arquivo DATASET-6.csv para o Colab antes de executar.'
    )

colunas_originais_6 = [
    'Global_active_power',
    'Global_reactive_power',
    'Voltage',
    'Global_intensity',
    'Sub_metering_1',
    'Sub_metering_2',
    'Sub_metering_3'
]

separador_6 = ';' if arquivo_dataset6.suffix.lower() == '.txt' else None
dados6_brutos = pd.read_csv(
    arquivo_dataset6,
    sep=separador_6,
    engine='python',
    na_values=['?', '']
)
inicio_dataset6 = ' '.join(
    map(str, dados6_brutos.head(2).to_numpy().ravel())
).lower()
if 'continuous' in inicio_dataset6 or 'meta' in inicio_dataset6:
    dados6_brutos = pd.read_csv(
        arquivo_dataset6,
        sep=separador_6,
        engine='python',
        skiprows=[1, 2],
        na_values=['?', '']
    )

dados6 = dados6_brutos[colunas_originais_6].copy()
for coluna in colunas_originais_6:
    dados6[coluna] = pd.to_numeric(dados6[coluna], errors='coerce')

linhas_antes_6 = len(dados6)
dados6.dropna(inplace=True)
linhas_removidas_6 = linhas_antes_6 - len(dados6)
dados6.reset_index(drop=True, inplace=True)

dados6.rename(columns={
    'Global_active_power': 'Potencia_Ativa',
    'Global_reactive_power': 'Potencia_Reativa',
    'Voltage': 'Tensao',
    'Global_intensity': 'Corrente',
    'Sub_metering_1': 'Submedicao_1',
    'Sub_metering_2': 'Submedicao_2',
    'Sub_metering_3': 'Submedicao_3'
}, inplace=True)

print(f'Linhas incompletas removidas: {linhas_removidas_6}')
print(f'Dimensões da amostra tratada: {dados6.shape[0]} linhas e {dados6.shape[1]} colunas')
dados6.head()
```

```python
dados6.info()
print()
print('Valores ausentes após o tratamento:')
display(dados6.isnull().sum().to_frame('Quantidade'))
dados6.describe()
```

## 2 e 3. Potência ativa máxima, limiar de 75% e primeiro DataFrame

```python
potencia_ativa_maxima_6 = dados6['Potencia_Ativa'].max()
limiar_75_6 = 0.75 * potencia_ativa_maxima_6
df_alta_potencia_6 = dados6[
    dados6['Potencia_Ativa'] > limiar_75_6
].copy()

print(f'Potência ativa máxima: {potencia_ativa_maxima_6:.3f} kW')
print(f'Limiar de 75%: {limiar_75_6:.3f} kW')
df_alta_potencia_6.sort_values(
    'Potencia_Ativa', ascending=False
).head(10)
```

## 4. Quantidade e percentual dos registros selecionados

```python
quantidade_alta_6 = len(df_alta_potencia_6)
percentual_alta_6 = quantidade_alta_6 / len(dados6) * 100

print(f'Registros acima do limiar: {quantidade_alta_6}')
print(f'Percentual da amostra: {percentual_alta_6:.4f}%')
```

## 5 e 6. Corrente média e segundo DataFrame

```python
corrente_media_6 = dados6['Corrente'].mean()
df_alta_potencia_corrente_6 = dados6[
    (dados6['Potencia_Ativa'] > limiar_75_6) &
    (dados6['Corrente'] > corrente_media_6)
].copy()

quantidade_duplo_6 = len(df_alta_potencia_corrente_6)
percentual_duplo_6 = quantidade_duplo_6 / len(dados6) * 100

print(f'Corrente média da amostra: {corrente_media_6:.3f} A')
print(f'Registros com as duas condições: {quantidade_duplo_6}')
print(f'Percentual da amostra: {percentual_duplo_6:.4f}%')
df_alta_potencia_corrente_6.head(10)
```

## 7. Comparação e interpretação

```python
reducao_6 = quantidade_alta_6 - quantidade_duplo_6

if reducao_6 == 0:
    efeito_corrente_6 = (
        'não reduziu a quantidade de registros, pois todos os episódios acima '
        'do limiar de potência também apresentaram corrente acima da média'
    )
else:
    efeito_corrente_6 = (
        f'reduziu a quantidade em {reducao_6} registros, mantendo somente os '
        'episódios que também apresentaram corrente acima da média'
    )

display(Markdown(f'''
**Interpretação:** o primeiro DataFrame possui **{quantidade_alta_6} registros**, e o segundo possui **{quantidade_duplo_6} registros**. A inclusão da corrente como segunda condição **{efeito_corrente_6}**.

Esse comportamento é coerente com a relação entre potência ativa e intensidade de corrente na instalação. Ainda assim, o resultado descreve a amostra analisada e deve ser interpretado junto de tensão, potência reativa e características dos equipamentos.
'''))
```
